````python
# 11. Tạo ngẫu nhiên 2 tensor kích thước x = (3, 2) và giá trị trong khoảng [5, 10], y = (2, 3) và giá trị [15, 20] tương ứng. Hãy giữ giá trị đường chéo của x và thực hiện x + y.
x = tf.random.uniform(shape=(3, 2), minval=5, maxval=10, dtype=tf.float32)
y = tf.random.uniform(shape=(2, 3), minval=15, maxval=20, dtype=tf.float32)
# Chuyển vị y để có kích thước (3, 2)
y_transposed = tf.transpose(y)
# Thực hiện phép cộng
z = x + y_transposed

# Lấy đường chéo của x
diag_x = tf.linalg.diag_part(x)
# Thay thế đường chéo của z bằng đường chéo của x
result = tf.linalg.set_diag(z, diag_x)
try:
  print("Tensor x:\n", x.numpy())
  print("Tensor y:\n", y.numpy())
  print("Tensor y sau khi chuyển vị:\n", y_transposed.numpy())
  print("Kết quả trước khi giữ đường chéo:\n", z.numpy())
  print("Kết quả cuối cùng:\n", result.numpy())
except tf.errors.InvalidArgumentError as e:
  print(f"Error: {e}\n")
  print("Tensor shapes are incompatible for addition.")
````

````python
# 12. Viết hàm tạo 2 tensor ngẫu nhiên có kích thước x = (3, 2, 2), y = (3, 2, 2). Viết hàm tính cộng,trừ, nhân, chia sau đó chuyển các hàm này vào trong graph.
# 1. Hàm tạo tensor ngẫu nhiên
def create_tensors():
    """Tạo 2 tensor x, y kích thước (3, 2, 2) với giá trị ngẫu nhiên [0, 1)"""
    x = tf.random.uniform(shape=(3, 2, 2))
    y = tf.random.uniform(shape=(3, 2, 2))
    return x, y

# 2. Định nghĩa các phép toán dưới dạng graph
@tf.function
def tensor_add(x, y):
  """Phép cộng phần tử"""
  return tf.add(x, y, name="Addition")

@tf.function
def tensor_sub(x, y):
  """Phép trừ phần tử"""
  return tf.subtract(x, y, name="Subtraction")

@tf.function
def tensor_mul(x, y):
  """Phép nhân phần tử"""
  return tf.multiply(x, y, name="Multiplication")

@tf.function
def tensor_div(x, y):
  """Phép chia phần tử"""
  return tf.divide(x, y, name="Division")

# 3. Thực thi và kiểm tra
if __name__ == "__main__":
  # Tạo tensor
  x, y = create_tensors()

  # Thực hiện các phép toán
  add_result = tensor_add(x, y)
  sub_result = tensor_sub(x, y)
  mul_result = tensor_mul(x, y)
  div_result = tensor_div(x, y)

  # In kết quả
  print("Input x:\n", x.numpy())
  print("\nInput y:\n", y.numpy())
  print("\nAddition result:\n", add_result.numpy())
  print("\nSubtraction result:\n", sub_result.numpy())
  print("\nMultiplication result:\n", mul_result.numpy())
  print("\nDivision result:\n", div_result.numpy())

print(tensor_add.pretty_printed_concrete_signatures())

def safe_div(x, y):
  return tf.where(y == 0, 0, x/y)
````

````python
# 13. Xây dựng model để tính tổng hai tensor x, y
# 1. Định nghĩa model
def build_sum_model(input_shape):
  """Xây dựng model tính tổng hai tensor cùng kích thước"""
  # Định nghĩa hai input layer
  input_x = tf.keras.Input(shape=input_shape, name='input_x')
  input_y = tf.keras.Input(shape=input_shape, name='input_y')

  # Layer cộng hai tensor
  sum_layer = tf.keras.layers.Add(name='addition_layer')([input_x, input_y])

  # Tạo model
  model = tf.keras.Model(
    inputs=[input_x, input_y],
    outputs=sum_layer,
    name='Sum_Model'
  )
  return model

# 2. Tạo và sử dụng model
if __name__ == "__main__":
  # Giả sử tensor có shape (2, 2) - có thể thay đổi theo yêu cầu
  input_shape = (2, 2)

  # Khởi tạo model
  model = build_sum_model(input_shape)

  # Tạo dữ liệu test
  x = tf.constant([[1., 2.], [3., 4.]])  # Shape (2,2)
  y = tf.constant([[5., 6.], [7., 8.]])  # Shape (2,2)

  # Dự đoán kết quả
  result = model.predict([x[tf.newaxis], y[tf.newaxis]])

  # In thông tin model và kết quả
  model.summary()
  print("\nInput x:\n", x.numpy())
  print("\nInput y:\n", y.numpy())
  print("\nSum result:\n", result)
````

````python
# 14. Xây dựng model để tính hiệu và tích hai tensor x, y
def diff_product_model(x, y):
  diff = tf.subtract(x, y)
  product = tf.multiply(x, y)
  return diff, product



diff_res, product_res = diff_product_model(x, y)
print("Difference:", diff_res)
print("Product:", product_res)
````
